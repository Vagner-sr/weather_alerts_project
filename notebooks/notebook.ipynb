```python
import requests
import pandas as pd
import json
```

```python
url = "https://geo.weather.gc.ca/geomet?service=WFS&version=2.0.0&request=GetFeature&typeNames=ALERTS&outputFormat=application/json"
response = requests.get(url)
data = response.json()

df = pd.json_normalize(data['features'])
df = df.to_csv('../data/raw/alerts_raw.csv', index=False)
```

```python
from sqlalchemy import create_engine
engine = create_engine("postgresql://postgres:211164@localhost:5432/weather_database")
df.to_sql('alerts', con=engine, if_exists='replace', index=False)
```
